```python
import numpy as np

list1 = [15.5, 25.11, 19.0]
list2 = [12.2, 1.3, 6.38] 

print('[list] Cat: ',list1 + list2)        # Concatenate two lists

print('[list] Sum:', end=' ')
for i in range(len(list1)): # Sum two lists
    print(list1[i] + list2[i], end=' ')
print()

array1 = np.array(list1)    # Create two 1-dimensional (1D) arrays
array2 = np.array(list2)    # with the elements of the above lists

print('[arr] Sum:', array1 + array2)      # Sum two 1D arrays
```

<section>
    <h1>Creating and modifying arrays</h1>
    <p>A NumPy array is an ordered, indexed, and mutable container.</p> 
    <p>NumPy provides many functions to create and modify arrays. Though mutable, NumPy arrays have a fixed shape, so any operation to change an array's shape actually creates a new version (a new Python object) of the array with the intended modification, leaving the original unchanged.</p>
    <p>If the elements in the array are unknown before creation, certain NumPy functions create placeholder values that can be changed during a program's run. 
    <extra>Ex: np.zeros((1, 3)) creates a (1, 3) shape array filled with 0s and np.ones((1, 3)) creates a (1, 3) shape array filled with 1s.</extra></p>
    <p>Some NumPy operations are instance methods of the object, Ex: array.sort(), while other NumPy operations are functions of the NumPy module, 
    <extra>Ex: np.delete(). This material takes an object-oriented approach and uses array methods. NumPy functions are used where array methods don't exist.</extra></p>
</section>

## Mathematical operations

Ex: [5 5 5] + [1 2 3] computes [5+1 5+2 5+3], or [6 7 8].

```python
# Figure 3.1.2: Common NumPy math operations and functions.
import numpy as np

array1 = np.array([10, 20, 30, 40])
array2 = np.array([1, 2, 3, 4])

# Some common array operations

print('Adding arrays (array1 + array2)')
print(array1 + array2)

print('\nSubtracting arrays (array1 - array2)')
print(array1 - array2)

print('\nMultiplying arrays (array1 * array2)')
print(array1 * array2)

print('\nCalculating dot product of arrays')
print(np.dot(array1, array2))

print('\nFinding square root of each element in array1')
print(np.sqrt(array1))

print('\nFinding minimum element in array1')
print(array1.min())

print('\nFinding maximum element in array1')
print(array1.max())
```
