## Indexed vs Unindexed

Indexed Arrays	
Non-Indexed Arrays

Indexing	
Integer indexes	
Key-based (string, hash, etc.)

Access	
Direct access using index	
Access using hash function or key

Operations	
Fast lookups, inserts, deletes	
Fast lookups, inserts, deletes, but may have collisions

indexed arrays 
data requiring direct access by index, 
such as matrices, vectors, or time series data.

Unindexed arrays 
data requiring key-based access, 
such as configuration files, JSON data, or caching mechanisms.

```python
my_non_indexed_array = {'a': 1, 'b': 2, 'c': 3}
print(my_non_indexed_array['b'])  # 2
```

```python
my_indexed_array = [1, 2, 3, 4, 5]
print(my_indexed_array[2]) # 3
```

## Mutable vs Immutable

Mutable Arrays	
Immutable Arrays

Modifiability	
Can be modified	
Cannot be modified

Resizability	
Can be resized	
Fixed size

Operations	
In-place operations	
Copy-on-write operations

```python
my_mutable_array = [1, 2, 3]
my_mutable_array.append(4)
print(my_mutable_array) # [1, 2, 3, 4]
```

```python
my_immutable_array = (1, 2, 3)  # Tuple in Python
try:
    my_immutable_array[0] = 4
except TypeError:
    print("Error: Cannot modify an immutable array")
```

mutable arrays 
data that needs frequent modification, 
such as dynamic data structures or user input.

immutable arrays 
data that should not be modified, 
such as constants, configuration data, or cached results.

## Unique vs Allow duplicate

```python
my_duplicate_array = [1, 2, 3, 2, 1]
print(my_duplicate_array)  # [1, 2, 3, 2, 1]
```

```python
my_unique_array = {1, 2, 3, 2, 1} # Set in Python
print(my_unique_array) # {1, 2, 3}
```

Unique Arrays	
Allow Duplicate Arrays 

Duplicates	
Automatically removes duplicates	
May contain duplicate elements

Implementation	
Set or hash table	
List or vector

Operations	
Fast membership testing	
Indexing, slicing
