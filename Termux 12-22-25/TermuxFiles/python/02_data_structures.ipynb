```python
int_array = [13, 7, 45, -3, 92, 17]

def FindMax(inputArray):
    max = inputArray[0]                 # Max Search Algorithm
    i = 1                               # ====================
    for i in range(len(inputArray)):    # FOR the length of an array,
        if (inputArray[i] > max):       # IF member is greater than max
            max = inputArray[i]         # THEN the max is equal to member.                       
    return max

print(FindMax(int_array))
```

<h2>Efficient algorithms and hard problems</h2>

<p>Computer scientists and programmers typically focus on using and designing efficient algorithms to solve problems. Algorithm efficiency is most commonly measured by the algorithm runtime, and an efficient algorithm is one whose runtime increases no more than polynomially with respect to the input size. However, some problems exist for which an efficient algorithm is unknown.</p>

<p><b>NP-complete</b> problems are a set of problems for which no known efficient algorithm exists. NP-complete problems have the following characteristics:</p>

<ul>
    <li>No efficient algorithm has been found to solve an NP-complete problem.</li>
    <li>No one has proven that an efficient algorithm to solve an NP-complete problem is impossible.</li>
    <li>If an efficient algorithm exists for one NP-complete problem, then all NP-complete problems can be solved efficiently.</li>
</ul>

<p>By knowing a problem is NP-complete, instead of trying to find an efficient algorithm to solve the problem, a programmer can focus on finding an algorithm to efficiently find a good, but non-optimal, solution.</p>

```python
array = [1, 2, 3 , 4]
print(array)
                                        # ArrayAppend Algorithm
def ArrayAppend(array, newItem):        # =====================
   currentSize = len(array)             # Determine current size
   array += [None]                      # Increase array size by 1.
   array[currentSize] = newItem         # Assigns the new item as last element.


ArrayAppend(array, 5)
print(array)
```

```python
my_list = [1, 2, 3 , 4]

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.length = 0

    def print_list(self):
        current = self.head
        while current:
            print(current.data)
            current = current.next
                                        # ListAppend Algorithm
    def list_append(self, new_node):    # ====================
        if self.head is None:           # Initializes list
            self.head = new_node        # list is empty
            self.tail = new_node        #
        else:                           # Assigned to new data:
            self.tail.next = new_node   # tails next pointer
            self.tail = new_node        # lists tail pointer
        self.length += 1                # Increase list length
    
my_list = LinkedList()

new_node1 = Node(1)
new_node2 = Node(2)
new_node3 = Node(3)
new_node4 = Node(4)

my_list.list_append(new_node1)
my_list.list_append(new_node2)
my_list.list_append(new_node3)
my_list.list_append(new_node4)
my_list.print_list()
```

<section>
<h2>Algorithms using data structures</h2>

<p>Some algorithms utilize data structures to store and organize data during the algorithm execution. Ex: An algorithm that determines a list of the top five salespersons, may use an array to store salespersons sorted by their total sales.</p>
</section>
<style>
    section {

    h2 {        
        font-size: 18pt;

```python
def display_top_five_salespersons(all_salespersons):
    # Create a list with 5 elements
    top_sales = [{"name": "", "sales_total": -1} for _ in range(5)]

    for salesperson in all_salespersons:
        # Check if salesperson's total sales is greater than the last top_sales element
        if salesperson["sales_total"] > top_sales[-1]["sales_total"]:
            # Assign the last element in top_sales with the current salesperson
            top_sales[-1] = {"name": salesperson["name"], "sales_total": salesperson["sales_total"]}
            # Sort top_sales in descending order
            top_sales.sort(key=lambda x: x["sales_total"], reverse=True)

    # Display the top five salespersons
    for salesperson in top_sales:
        print(salesperson)
```

# **(2.5)** *Abstract data types*

<section>
<h2>Abstract data types (ADTs)</h2>

<p>An abstract data type (ADT) is a data type described by predefined user operations, such as "insert data at rear," without indicating how each operation is implemented. An ADT can be implemented using different underlying data structures. However, a programmer need not have knowledge of the underlying implementation to use an ADT.</p>

<p><i>Ex: A list is a common ADT for holding ordered data, having operations like append a data item, remove a data item, search whether a data item exists, and print the list. A list ADT is commonly implemented using arrays or linked list data structures.</i></p>
</section>

<p>Abstraction means to have a user interact with an item at a high-level, with lower-level internal details hidden from the user. ADTs support abstraction by hiding the underlying implementation details and providing a well-defined set of operations for using the ADT.</p>
