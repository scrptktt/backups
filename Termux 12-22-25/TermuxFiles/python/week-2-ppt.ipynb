<h3><u>Primitives</u></h3>
<ul>
    <li>uses simple data types.</li>
    <li>has a known size.</li>
    <li>variable values exists on stack.</li>
    <li>retrieves value on one lookup.</li>
</ul>
<h3><u>Objects</u></h3>
<ul>
    <li>uses complex data types.</li>
    <li>has an unpredeterminable size.</li>
    <li>variable values exists on heap.</li>
    <li>retrieves value on two lookups.</li>
</ul>
</div>

<br></br>
<!-- [2x2] COMPARE BOX: data types -->
<div>
<h3><u>STACKS</u></h3>
<ul>
    <li>Temporary</li>
    <li>Fast</li>
    <li>Limited storage</li>
    <li>Continous memory</li>
    <li>Automatic allocation</li>
    <li>Automatic deallocation</li>
</ul>
<h3><u>HEAPS</u></h3>
<ul>
    <li>also has reference on stack.</li>
    <li>Slower.</li>
    <li>Dynamic (keep as long as needed).</li>
    <li>Manual allocation</li>
    <li>Manual deallocation</li>
</ul>
</div>

```python
class MyClass: # Sign, Pass
    def __init__(self, value):
        self.value = value      # Instance variable stored in heap

def create_object():
    obj = MyClass(10)           # Object created in heap
    print(obj.value)

create_object()
```
